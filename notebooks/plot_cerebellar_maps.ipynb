## Plot cerebellar activity maps 

```python
# Load data 
base_dir = ut.get_base_dir()
X,info,myds = ds.get_dataset(base_dir,'Social',atlas='MNISymC3',subj='group',type='CondAll',sess='ses-social')
X = X[0]
```

```python
X = X-X.mean(axis=0)
plt.figure(figsize=(12,17))
for i in range(X.shape[0]):
    plt.subplot(5,4,i+1)
    fpl.plot_cerebellum(X[i,:],new_figure=False,cscale=[-0.25,0.25])
    plt.title(info.names[i])
plt.tight_layout()
```
